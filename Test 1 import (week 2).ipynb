```python
import h5py
import numpy as np
import matplotlib.pyplot as plt

# Load ShapeSpace.mat using h5py
with h5py.File('ShapeSpace.mat', 'r') as f:
    shape_data = f['ShapeSpace'][:]        # shape: (50, 50, 248396)
    shape_data = np.transpose(shape_data, (2, 0, 1))  # reshape to (248396, 50, 50)

# Load PropertySpace.mat using h5py
with h5py.File('PropertySpace.mat', 'r') as f:
    property_data = f['PropertySpace'][:]  # shape: (5, 248396)
    property_data = property_data.T        # reshape to (248396, 5)

# Confirm shape
print("ShapeSpace:", shape_data.shape)
print("PropertySpace:", property_data.shape)

# Show one example
index = 0
plt.imshow(shape_data[index], cmap='gray')
plt.title("Microstructure Example")
plt.axis('off')
plt.show()

print("C11, C12, C22, C66, VF =", property_data[index])
```
